Create a Schema my_coffee_shop

```sql
--create schema my_coffee_shop

CREATE SCHEMA IF NOT EXISTS my_coffee_shop
COMMENT 'Bright coffee shop Analytics';
USE my_coffee_shop;
SELECT current_database(); -- confirm: should return my_coffee_shop
```

Create table named dim_customers with columns

customer_id(PK)
first_name
last_name
email
tier
loyalty_points

```sql
--create table name dim_customers

CREATE TABLE dim_customers (
    customer_id INT PRIMARY KEY,
    first_name STRING NOT NULL,
    last_name STRING NOT NULL,
    email STRING,
    tier STRING,
    loyalty_points INT,
    created_at TIMESTAMP
);
```

```sql
-- confirm
SELECT * FROM my_coffee_shop.dim_customers;
```

Create the dim_products table.


```sql
--create table name dim_product

CREATE TABLE IF NOT EXISTS dim_products (
 product_id INT PRIMARY KEY,
 product_code STRING,
 product_name STRING,
 category STRING,
 price DECIMAL (10, 2),
 is_active BOOLEAN
);
```

```sql
-- confirm
SELECT * FROM my_coffee_shop.dim_products;
```

```sql
-- confirm
SELECT * FROM my_coffee_shop.dim_products;
```

Create the fact_orders table, partitioned by order_date.

```sql
---create table fact_orders using delta partitioned by order_date

CREATE TABLE fact_orders (
    order_id BIGINT PRIMARY KEY,
    customer_id INT,
    product_id INT,
    store_id INT,
    order_date DATE,
    quantity INT,
    unit_price DECIMAL (8, 2),
    total_amount DECIMAL (10, 2)
)
USING DELTA 
PARTITIONED BY (order_date)
```

```sql
-- confirm
SELECT * FROM my_coffee_shop.fact_orders;
```

Enable column mapping on dim_customers.


```sql
--alter table by enabling column mapping, using set tblproperties 

ALTER TABLE my_coffee_shop.dim_customers SET TBLPROPERTIES('delta.columnMapping.mode'='name') 
```

Update all Bronze-tier customers: add 50 to their points.
REQUIRED STEPS: SELECT (test WHERE first) UPDATE … SET … WHERE
★ Step 1: run SELECT with your WHERE clause and confirm the right rows appear. Step 2: run the UPDATE.
Step 3: run SELECT again to confirm the change.


```sql
-- Step 1: test your WHERE clause
SELECT customer_id, first_name, tier, points
FROM my_coffee_shop.dim_customers
WHERE tier = 'Bronze';
```

```sql
-- Step 2: run the UPDATE
UPDATE my_coffee_shop.dim_customers
SET points = points + 50
WHERE tier = 'Bronze';
```

```sql
-- Step 3: confirm
SELECT customer_id, first_name, tier, points
FROM my_coffee_shop.dim_customers
WHERE tier = 'Bronze';
```

Back up dim_customers, then delete one customer by their customer_id.
REQUIRED STEPS: CREATE TABLE … AS SELECT SELECT (verify row) DELETE … WHERE
customer_id = ? DESCRIBE HISTORY
★ Always back up before deleting. Delete by customer_id only - never by name. Check DESCRIBE HISTORY to
see the delete in the Delta log.


```sql
-- Step 1: back up first
CREATE TABLE my_coffee_shop.dim_customers_backup AS SELECT * FROM my_coffee_shop.dim_customers;
```

```sql
-- Step 2: Locate and verify the row you are about to delete
SELECT * FROM my_coffee_shop.dim_customers WHERE customer_id = 3;
```

```sql
-- Step 3: delete by ID only
DELETE FROM my_coffee_shop.dim_customers WHERE customer_id = 3;
```

```sql
-- Step 4: confirm
SELECT * FROM my_coffee_shop.dim_customers;
```

```sql
-- Step 5: check the Delta log
DESCRIBE HISTORY my_coffee_shop.dim_customers;

```
